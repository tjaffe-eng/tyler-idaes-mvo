# Getting Started

## Git terminology

### Summary
**branch**
A name for a series of commits. See [Branches](#branches).

**fork**
Copy of a repository in GitHub. See [Forks](#forks).

**pull request (PR)**
A request to compare and merge code in a GitHub repository. See [Pull Requests](#pull-requests).

#### Branches
A branch is a series of commits that allows you to separate the code development from the main code. There is a good description of what Git branches are and how they work here. Understanding this takes a little study, but this pays off by making Git’s behavior much less mysterious. The short, practical version is that a branch is a name for a series of commits that you want to group together, and keep separable from other series of commits. From Git’s perspective, the branch is just a name for the first commit in that series.

It is recommended that you create new branches on which to develop your work, and reserve the “main” branch for merging in work that has been completed and approved on GitHub. One way to do this is to create branches that correspond directly to issues on GitHub, and include the issue number in the branch name.

#### Forks
A fork is a copy of a repository, in the GitHub shared space (a copy of a repository from GitHub down to your local disk is called a “clone”). In this context, that means a copy of the “idaes-dev” repository from the IDAES organization (IDAES/idaes-dev) to your own user space, e.g., myname/idaes-dev). The mechanics of creating and using forks on GitHub are given here.

#### Pull requests
A fundamental procedure in the development lifecycle is what is called a “pull request”. Understanding what these are, and do, is important for participating fully in the software development process. First, understand that pull requests are for collaborative development (GitHub) and not part of the core revision control functionality that is offered by Git. The official GitHub description of pull requests is here. However, it gets technical rather quickly, so a higher-level explanation may be helpful:

Pull requests are a mechanism that GitHub provides to look at what the code on some branch from your fork of the repository would be like if it were merged with the main branch in the main (e.g., idaes-pse/idaes-dev) repository. You can think of it as a staging area where the code is merged and all the tests are run, without changing the target repository. Everyone on the team can see a pull request, comment on it, and review it.

Both Git and GitHub are widely used and there are excellent tutorials and resources for each. See See [Atlassian Github tutorials](https://www.atlassian.com/git/tutorials/git-with-github), [GitHub help](https://docs.github.com/en), and [Git documentation](https://git-scm.com/doc).

#### Installation with GitHub

The main IDAES-MVO GitHub repository is idaes-mvo. This repository includes the core framework, python packages and integrated tools for solving process family design problems. It contains all of the release versions of IDAES-MVO.

The following instructions describe how to install and update the idaes-mvo repository.

#### Github setup

In order to use GitHub, you need to create a login on [Github](https://github.com/).

#### Fork the repository

You use a “fork” of a repository (or “repo” for short) to create a space where you have complete control and can make changes without directly affecting the main repository.

You should first visit the idaes-mvo repo on Github at [IDAES/idaes-mvo](https://github.com/IDAES/idaes-mvo). Then you should click on the fork icon in the top right and click on your username. These steps will have created your own fork of the repo with the same name under your username.

#### Clone your fork

A “clone” is a copy of a Github repository on your local machine. This is what you need to do in order to actually edit and change the files. To make a clone of the fork you created in the previous step, change to a directory where you want to put the source code and run the command:
```bash
git clone https://github.com/MYNAME/idaes-pse.git
cd idaes-pse
```
Of course, replace MYNAME with your username. This will download all the files in the latest version of the repository onto your local disk.

```{note}
After the git clone, subsequent Git commands should be performed from the “idaes-mvo” directory.
```

#### Add upstream remote

In order to guarantee that your fork can be synchronized with the “main” idaes-mvo repo in the GitHub IDAES organization, you need to add a pointer to that repository as a remote. This repository will be called upstream and linked with the following command:
```bash
git remote add upstream https://github.com/IDAES/idaes-pse.git
```
To check to see if you added the remote correctly use the following command:

```bash
git remote -v
```
You should see that there are two remotes, origin and upstream. Both have two lines showing the remote name, the url, and the access (fetch or push). Origin is the pointer to your fork and was automatically added with the clone command, while upstream is the pointer to the main idaes-pse repo that you just added.

### Create the Python environment
Once you have the repo cloned, you can change into that directory (by default, it will be called “idaes-mvo” like the repo) and install the Python packages.

But before you do that, you need to get the Python package manager fully up and running. We use a Python packaging system called [Conda](https://docs.conda.io/en/latest/) and we specifically use a community driven minimal version [Miniforge](https://conda-forge.org/miniforge/).

After Miniforge is installed, we recommend creating a separate conda environment for IDAES-MVO. Create and activate a conda environment for the new IDAES-MVO installation with the following commands (we officially support python 3.13):

```bash
conda create --name idaes-mvo python=3.13
conda activate idaes-mvo
```
```{note}
When setting up a conda environment like this, you must conda activate idaes-mvo whenever you open a fresh terminal window and wish to use IDAES-MVO.
```

#### Finish the installation

Now that conda and pip are installed, and you are in the “idaes-mvo” conda environment, you can run the following commands to install the requirements for IDAES-MVO (e.g. binaries for the solver CBC and other external function calls):

```bash
pip install -r requirements-dev.txt
```
```{note}
This pip install command would override any package within the conda environment, so if you would like to use a specific version of a package (e.g. a local clone of the Pyomo git repository), you should look at the requirements-dev.txt file and use it as a reference to either install the individual packages manually, or create a separate requirements file customized to your development use case.
```

Once installed, the example scripts in the repository can be used to reproduce the two main workflows:

- Discretized approach
- Machine learning surrogate approach

The examples directory is the best place to start for a practical walkthrough.

## API reference

The package is organized around a small set of modules:

- `process_family` for the overall package interface
- `process_family.type` for model and problem-type definitions
- `process_family.utils` for utility functions

These modules are implemented in the repository source tree and can be inspected directly in the package directories:

- process_family
- process_family/type
- process_family/utils

For users who want a quick overview, the examples in the repository provide the most practical entry points for the discretized and surrogate workflows.

## Git commands
The Git tool has many different commands, but there are several really important ones that tend to get used as verbs in software development conversations, and therefore are good to know:

**add**
Put a file onto the list of “things I want to commit” (see “commit”), called “staging” the file.

```bash
git add <file-name>
```
**commit**
Save the changes in “staged” files into Git (since the last time you did this), along with a user-provided description of what the changes mean (called the “commit message”).

```bash
git commit -m "<Description of what is being added to or changed about the repository>"
```
**push**
Move local committed changes to the GitHub-hosted “remote” repository by “pushing” them across the network.

```bash
git push <remote-name> <branch-name>
```
**pull**
Update your local files with changes from the GitHub-hosted “remote” repository by “pulling” them across the network.

```bash
git pull <remote-name> <branch-name>
```
```{note}
The push and pull commands require GitHub (or some other service that can host a remote copy of the repository).
```
