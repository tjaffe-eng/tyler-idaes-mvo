# Examples

The repository contains example python scripts and notebooks that show how the two modeling approaches are applied.

## Discretized approach workflow

The **discretization approach** simplifies the design process by choosing from a predefined list of design options (like standard equipment sizes) instead of dealing with infinite possibilities. This shortcut transforms a highly complex, slow-to-solve optimization problem with yes/no decisions and continuouse physical curves, into a much faster, linear one called a Mixed-Integer Linear Program (MILP) {cite:p}`stinchfield2024mixed`. Crucially, it does this while still capturing the core trade-offs between using standardized, shared parts and adding customized features.

This makes it a perfect starting point if you want to understand the basics of process family design before diving into advanced machine learning (ML) surrogate models. It is also your best option when data is scarce, since training highly accurate ML models typically requires a massive dataset that you might not have access to yet.

*   **Interactive Tutorial:** Walk through the step-by-step discretized.ipynb notebook in the `example` folder. 


## Machine learning surrogate approach workflow

![Machine Learning Surrogate Approach](_static/papers/MILP-representable_ML_Surrogates_approach.png)

The **surrogate-based approach** uses machine learning (ML) models to mimic and simplify complex, highly curved physical equations (like fluid dynamics or chemical reactions). By embedding these smart "shortcut models" (surrogates) directly into our optimization framework, we can search the entire continuous design space comprehensively. This swaps out slow, computationally expensive physics calculations for fast, straight-line approximations (specifically, piecewise-linear constraints) {cite:p}`stinchfield2025mixed`. 

In practice, this means we take the incredibly complex physics equations and replace them with these machine learning predictions. This trick keeps the structural logic of the design framework completely intact (known mathematically as Generalized Disjunctive Programming, or GDP) but makes it vastly faster to solve {cite:p}`stinchfield2025mixed`.
