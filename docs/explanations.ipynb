# Explanations

## Why Multi-Variant Optimization?

Process systems engineering has traditionally focused on optimizing a single process instance for a specific deployment context. That approach is powerful, but it does not address the manufacturing and deployment challenges that arise when many similar systems must be designed and installed at scale.

The central idea of process family design is to treat a collection of related process systems as one coordinated design problem. Instead of designing each deployment independently, the formulation identifies common components and shared platform elements while still allowing some customization where it matters.

This perspective is especially relevant for technologies such as carbon capture, desalination, and other modular energy systems where rapid deployment and cost-effective manufacturing are essential.

## Core Concepts

This repository is structured to systematically address the following core objectives:

- Standardization of shared equipment or design choices
- Customization of site-specific or deployment-specific elements
- Optimization of both capital and deployment costs
- Use of model reduction strategies to produce accurate solutions within shorter exectuion times

The resulting product should closely resemble the image below, which depicts both a typical process platform and process family.

![Process Family Design Image](_static/papers/process_family_design.png)

The motivation behind this work is that significant savings can arise from economies of numbers and from designing families of processes rather than one-off installations. For further reading and a deeper explanation of this framework, see the four papers referenced in the [Further Reading](#further-reading) section.

## Definitions and Terminology 

### Key Definitions

![Process Architecture Image](_static/papers/Process_System_Architecture.png)

<span style="font-size: 1.3em;">**Process System Architecture**</span>: The complete set of general unit operations (e.g., compressors, valves) needed to build the entire process family.

<span style="font-size: 1.3em;">**Unit Module Type**</span>: A general component category within the architecture (e.g., "feed preheater") that can be either standardized or uniquely designed.

<span style="font-size: 1.3em;">**Unit Module Design**</span>: A specific, engineered instance of a module type (e.g., a shell-and-tube heat exchanger with a 50 $m^2$ area).

![Process Variant Image](_static/papers/Process_Variants.png)

<span style="font-size: 1.3em;">**Process Variant**</span>: A specific process design within a family, defined by unique input parameters and performance constraints.

![Process platform and process family Image](_static/papers/process_platform_and_process_family.png)

<span style="font-size: 1.3em;">**Process Platform**</span>: The core catalog of standardized unit module designs shared across multiple process variants.

<span style="font-size: 1.3em;">**Process Variant Design**</span>: A specific configuration combining shared platform designs, custom unique designs, and optimized operating variables.

<span style="font-size: 1.3em;">**Process Family**</span>: The entire collection of process variant designs that successfully meet all system requirements.

## Strucutre and Components

### Package Structure
The package is organized around a small set of modules:

- `process_family` for the overall package interface
- `process_family.type` for model classes for both discretized and ML surrogate approaches
- `process_family.utils` for utility functions and machine learning tools

These modules are implemented in the repository source tree and can be inspected directly in the package directories:

- process_family
- process_family/type
- process_family/utils

### Components

#### Utility Classes

The **Parameters** Class Creates a Parameter object, that holds all of the information we need to build any of the 
process family design optimization formulations.

```{note}
This was designed to have all attributes added in any order.
Some more logical checks are performed when a process_family.type object is built.
```

```bash
params = Parameters("<data_path>")
```
The **SurrogateParameters** Class creates the Parameter object for the MILP-representable ML Surrogates approach class, which is a child of the base class **Parameters** holds all of the information we need to build MILP-representable ML Surrogates process family design optimization formulation.

```bash
params = SurrogateParameters("<data_path>")
```

#### Machine Learning Training Classes

The **BaseTrainer** class initializes the base information for training ML surrogates for process family design. Organizes and splits the training, testing, and validation data.

The **NNTrainer** class is contains the tools for training neural networks to construct peice-wise linear surrogate models to approaximate the nonlinear systems of equations that arise within the process family design problem. It is also a child class of the parent class **BaseTrainer**.

```bash
nn_trainer = NNTrainer(params = params, 
                       label = "<data_path>")
```

The **TreeTrainer** class contains the tools for training trees (e.g. gradient boosted decision tree & linear model decision tree) to construct peice-wise linear surrogate models to approaximate the nonlinear systems of equations that arise within the process family design problem. It is also  a child class of the parent class **BaseTrainer**.

```bash
tree_trainer = TreeTrainer(params = params, 
                           label = "<data_path>")
```

#### Optimization Modeling Classes

The **ProcessFamilyBase** class is the parent class for all other ProcessFamily types ( e.g. discretization and surrogates).
It organizes the specified data into the necessary sets used across the different process family design optimization techniques

The **DiscretizedProcessFamily** class is the discretized approach class, which is a child of the base class **ProcessFamilyBase**. It builds the equivalent Pyomo model for the discretized optimization formulation  of the process family design problem.

```bash
dpfd = DiscretizedProcessFamily(params)
```
The **SurrogatesProcessFamily** is the MILP-representable ML Surrogates approach class, which is a child of the base class **ProcessFamilyBase**. It builds the equivalent Pyomo model for the MILP-representable ML Surrogates optimization forumlation of the process family design problem.

```bash
spfd = SurrogatesProcessFamily(params)
```
